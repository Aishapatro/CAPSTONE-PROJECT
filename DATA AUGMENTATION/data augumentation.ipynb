```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img, img_to_array

#create an image data generator with augementation parameters
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    rescale=None,
    fill_mode='reflect'  #nearest, reflect, wrap, constant
)

img=load_img(r"C:\Users\aisha\OneDrive\Documents\ME.jpeg")  # Load the image
```

```python
img
```

```python
x=img_to_array(img)  # Convert the image to a numpy array
x=x.reshape((1,)+x.shape)  # Reshape the array to (1, height, width, channels)

i=0
for batch in datagen.flow(x, batch_size=1, 
                          save_to_dir=r"C:\Users\aisha\OneDrive\Desktop\DATA AUGMENTATION\ME.jpeg", 
                          save_prefix='ME', save_format='jpeg'):
    i+=1
    if i>30:
        break  # Generate 20 augmented images
```
